# Fase 3: Procesar

¿Qué herramientas se eligieron y por qué?

Elegí Excel para la limpieza inicial porque permite procesar los archivos de manera secuencial y estructurada, ademas me ayudo a entender como estaban estructurados los datos. Esta herramienta facilita la inspección visual inmediata de los errores de formato (como marcas de tiempo corruptas) y ofrece funciones directas, como Tablas Dinámicas y "Texto en columnas", para auditar y transformar los datos de forma controlada antes de proceder al análisis masivo.
Despues se utilizó Python para hacer un limpieza sobre todos los archivos, esto para darle una clasificación a celdas que no contenian datos, pero solo en un numero de columnas determiando que se explica mas adelante. 

¿Qué medidas se tomaron para garantizar que los datos estén limpios?
* Eliminación de registros duplicados utilizando el identificador único de cada viaje.
* Estandarización del formato de fechas y horas a dd/mm/aaaa hh:mm:ss.
* Transformación de las coordenadas geográficas de texto a valores numéricos mediante la eliminación de apóstrofes ocultos y ajuste de separadores decimales.
* Creación de las variables calculadas ride_length (con formato de acumulación de horas) y day_of_week (estandarizada a números enteros).
* Identificación de datos faltantes en cada columna mediante la función CONTAR.BLANCO para su correcta gestión.

**Comprobación de datos faltantes:**

Para ver las celdas sin datos se utilizó la siguiente formula:

De esta forma se pudo comprobar que en cada unos de los 12 archivos, habia datos faltantes en las mismas 6 columnas.

**Despues se agregaron las siguientes dos columnas:**

    Que se calculó restando la columna started_ad de finished_ad.

* day_of_week: dia de la semana en que inicio el viaje.

    Se calculo usando la formula =DIASEM(). Por ejemplo:
    =DIASEM(C2, 1)
    El primer argumento de la formula hace referencia al dato de la columna started_ad y el segundo al formato de la semana (1 = domingo y 7 = sabado).

Antes de iniciar esta fase se ejecutó un script de Python con el proposito de clasificar las celdas que estan vacias dentro de las columnas que guardan el ID y el Nombre de las estaciones de esta forma: 

* star_station_name = las celdas que este vacias se rellenaran con "unknown"
* star_station_id = las celdas que este vacias se rellenaran con "NULL"
* end_station_name = las celdas que este vacias se rellenaran con "unknown"
* end_station_id = las celdas que este vacias se rellenaran con "NULL"

```python
import pandas as pd
import glob
import os

# /kaggle/input/datasets/abnerruiz/cyclistic-case-study-2025-2026

# Ruta del dataset
ruta_archivos = "/kaggle/input/datasets/abnerruiz/cyclistic-case-study-2025-2026/*.xlsx"
archivos = glob.glob(ruta_archivos)

if len(archivos) == 0:
    print("Error: Python no encontró ningún archivo. Verifica que la ruta del dataset sea correcta.")
else:
    print(f"¡Éxito! Python encontró {len(archivos)} archivos. Iniciando limpieza...\n")
    
    # Crea la carpeta de salida en el directorio de trabajo de Kaggle
    carpeta_salida = "/kaggle/working/archivos_procesados"
    os.makedirs(carpeta_salida, exist_ok=True)

    for archivo in archivos:
        nombre_original = os.path.basename(archivo)
        print(f"Procesando: {nombre_original}...")
        
        # Leer el archivo
        df = pd.read_excel(archivo)
        
        # Aplicar tus reglas de limpieza
        if 'star_station_name' in df.columns:
            df['star_station_name'] = df['star_station_name'].fillna('unknown')
            df['star_station_id'] = df['star_station_id'].fillna('NULL')
        elif 'start_station_name' in df.columns:
            df['start_station_name'] = df['start_station_name'].fillna('unknown')
            df['start_station_id'] = df['start_station_id'].fillna('NULL')
            
        df['end_station_name'] = df['end_station_name'].fillna('unknown')
        df['end_station_id'] = df['end_station_id'].fillna('NULL')
        
        # Se conserva exactamente el mismo nombre del archivo original
        ruta_guardado = os.path.join(carpeta_salida, nombre_original)
        df.to_excel(ruta_guardado, index=False)
        
    print(f"\n¡Proceso finalizado con éxito! Los archivos limpios están en '{carpeta_salida}'.")
```

# Fase 4: Analizar

Con los datos limpios, el objetivo de esta fase es descubrir las tendencias de uso y comprender cómo se comportan los diferentes tipos de ciclistas a lo largo del año. Para responder a la pregunta de negocio principal, el análisis se centra en identificar patrones de estacionalidad y comparar el volumen de viajes.

A través de Python (`pandas`, `seaborn` y `matplotlib`), se procesa el conjunto de datos masivo para agrupar métricas temporales clave y generar visualizaciones claras. El análisis explora:
* **La estacionalidad del servicio:** Cómo fluctúa la demanda de bicicletas mes a mes.
* **El volumen por usuario:** La comparación directa en la cantidad total de viajes entre miembros anuales y ciclistas ocasionales a lo largo de los últimos 12 meses.

**1. Preparación del Entorno y Localización de Datos**

El primer paso del análisis consiste en configurar el entorno de trabajo importando las librerías esenciales:
* **Manipulación y sistema:** `pandas` para el manejo de los datos, junto con `os` y `glob` para automatizar la lectura de múltiples archivos en el directorio.
* **Visualización:** `matplotlib.pyplot` y `seaborn` para trazar los gráficos estadísticos.

Una vez cargadas las herramientas, se define la ruta de la carpeta que contiene los archivos procesados y se rastrean todos los documentos con extensión `.xlsx`. Por último, se inicializa la lista vacía `lista_dataframes`, la cual funcionará como un contenedor temporal para almacenar los datos de cada mes antes de consolidarlos en un único conjunto de datos.

```python
import pandas as pd
import glob
import os
import matplotlib.pyplot as plt
import seaborn as sns

ruta_archivos = '/kaggle/working/archivos_procesados'
todos_los_archivos = glob.glob(os.path.join(ruta_archivos, "*.xlsx"))

lista_dataframes = []
```

**2. Integración, Transformación y Limpieza de Datos**

En este bloque, se procesa cada archivo mensual de forma individual para aplicar transformaciones críticas de calidad antes de consolidarlos. Las acciones específicas por archivo incluyen:

* **Conversión de Fechas:** Las columnas `started_at` y `ended_at` se estandarizan al formato `datetime`.
* **Cálculo y Filtrado de Duración (`ride_length`):** Se calcula el tiempo exacto de cada viaje. Para garantizar la confiabilidad del análisis, se descartan los registros anómalos con duraciones negativas o iguales a cero (como errores de sistema o bicicletas extraídas y devueltas al instante).
* **Extracción del Día (`day_of_week`):** Se deriva el día de la semana de forma numérica para facilitar la evaluación de tendencias diarias.
* **Deduplicación:** Se eliminan los viajes duplicados basándose en el identificador único del recorrido (`ride_id`).

Finalmente, los 12 meses filtrados se concatenan en el *DataFrame* maestro (`df_anual`), reportando en pantalla el volumen total de registros válidos listos para la exploración estadística.

```python
for archivo in todos_los_archivos:
    df = pd.read_excel(archivo)
    
    df['started_at'] = pd.to_datetime(df['started_at'])
    df['ended_at'] = pd.to_datetime(df['ended_at'])
    
    df['ride_length'] = df['ended_at'] - df['started_at']
    df = df[df['ride_length'].dt.total_seconds() > 0]
    
    df['day_of_week'] = (df['started_at'].dt.dayofweek + 1) % 7 + 1
    df = df.drop_duplicates(subset=['ride_id'])
    
    lista_dataframes.append(df)
    print(f"Procesado: {os.path.basename(archivo)} - Filas válidas: {len(df)}")

df_anual = pd.concat(lista_dataframes, ignore_index=True)
print(f"\nConsolidación exitosa. Total de registros anuales: {len(df_anual)}")
```

Nota: En total se analizan 5,932,320 filas en total.

**3. Análisis Macro: Volumen y Proporción de Usuarios**

Este bloque calcula la participación de mercado de cada segmento, determinando el volumen exacto de viajes y el porcentaje total que representan los miembros anuales frente a los usuarios casuales. Esta métrica base es fundamental para dimensionar el tamaño de la oportunidad comercial para la nueva estrategia de marketing.

```python
print("\n" + "="*50)
print("ANÁLISIS MACRO: MÉTRICAS CLAVE")
print("="*50)

# 1. Volumen total y porcentaje de viajes
print("\n--- 1. Volumen de Viajes por Segmento ---")
conteo_usuarios = df_anual['member_casual'].value_counts()
porcentaje_usuarios = df_anual['member_casual'].value_counts(normalize=True) * 100

for tipo, total in conteo_usuarios.items():
    print(f"{tipo.capitalize()}: {total:,.0f} viajes ({porcentaje_usuarios[tipo]:.2f}%)")
```

**4. Estadísticas de Tiempo de Uso (Duración de Viajes)**

Para entender las diferencias en los hábitos de uso, este bloque extrae estadísticas descriptivas clave sobre la duración de los recorridos para cada segmento.

* **Conversión a minutos:** Transforma la variable temporal (`ride_length`) a un valor numérico continuo (minutos) dividiendo los segundos totales entre 60.
* **Agrupación y métricas:** Utiliza la función `groupby` para separar a los miembros anuales de los casuales y aplica el método `agg` para calcular simultáneamente cuatro indicadores:
  * **Promedio (`mean`):** La duración típica general de un viaje.
  * **Mínimo y Máximo (`min`, `max`):** Los extremos inferior y superior del comportamiento de viaje.
  * **Mediana (`median`):** El punto central exacto de los datos. Esta métrica es vital en este análisis, ya que evita que viajes atípicamente largos sesguen la percepción del tiempo real que el usuario promedio pasa en la bicicleta.

```python
# 2. Estadísticas de tiempo (en minutos)
print("\n--- 2. Estadísticas de Tiempo de Uso (en minutos) ---")
df_anual['ride_length_mins'] = df_anual['ride_length'].dt.total_seconds() / 60
stats_duracion = df_anual.groupby('member_casual')['ride_length_mins'].agg(['mean', 'min', 'max', 'median'])
print(stats_duracion.round(2))
```

**5. Frecuencia de Viajes por Día de la Semana**

Para identificar si existen patrones diferenciados entre días laborables y fines de semana, este bloque contabiliza el volumen de viajes distribuidos por cada día.

* **Agrupación y conteo:** Utiliza `groupby` combinado con `size()` para contar el total de recorridos cruzando dos variables fundamentales: el tipo de usuario y el día de la semana.
* **Transformación matricial:** Aplica la función `unstack()` para pivotar los resultados y transformarlos en una tabla de doble entrada. En esta matriz, las filas representan el perfil del ciclista y las columnas los días de la semana, facilitando la identificación rápida de los picos de mayor demanda para cada segmento.

```python
# 3. Viajes por día de la semana
print("\n--- 3. Viajes por día de la semana ---")
uso_dias = df_anual.groupby(['member_casual', 'day_of_week']).size().unstack()
print(uso_dias)
```

**6. Preferencia de Tipo de Bicicleta por Segmento**

Este bloque examina las preferencias de equipamiento para descubrir si los miembros y los ciclistas ocasionales eligen diferentes tipos de bicicletas (clásicas, eléctricas o ancladas).

* **Agrupación cruzada:** Utiliza `groupby` para cruzar la categoría del usuario con el tipo de vehículo (`rideable_type`), generando un conteo absoluto de viajes para cada combinación.
* **Cálculo de distribución interna:** Mediante un ciclo `for`, el código extrae el conteo de cada bicicleta y lo divide entre el volumen total de viajes de ese grupo de usuarios específico (utilizando la variable `conteo_usuarios` calculada previamente).
* **Porcentaje relativo:** El resultado se multiplica por 100 y se formatea para mostrar la proporción exacta de uso. Esto revelará, por ejemplo, si los ciclistas casuales tienen una mayor afinidad por las bicicletas eléctricas que los usuarios anuales.

```python
# 4. Preferencia de tipo de bicicleta por usuario
print("\n--- 4. Uso de Tipo de Bicicleta por Usuario ---")
conteo_bicis = df_anual.groupby(['member_casual', 'rideable_type']).size()

for (usuario, bici), conteo in conteo_bicis.items():
    total_usuario = conteo_usuarios[usuario]
    porcentaje = (conteo / total_usuario) * 100
    print(f"{usuario.capitalize()} - {bici}: {porcentaje:.2f}%")
```

**7. Visualización: Serie de Tiempo Mensual (Estacionalidad)**

Para ilustrar el impacto de las estaciones del año en la demanda del servicio, este bloque genera un gráfico de líneas que compara el volumen de viajes mensual entre ambos segmentos.

* **Preparación Temporal:** Se extrae el año y mes (`YYYY-MM`) de la fecha de inicio usando `to_period('M')`. Posteriormente, se agrupan los datos para calcular el total de viajes por mes y por tipo de usuario.
* **Configuración Estética:** Se establece un lienzo amplio de 14x7 pulgadas con un fondo limpio (`whitegrid`) para facilitar la lectura.
* **Trazado de Tendencias:** Utilizando la librería Seaborn (`lineplot`), se grafican las series de tiempo con marcadores visuales en cada punto de datos. Se define una paleta de colores contrastante: naranja para ciclistas casuales y azul para miembros anuales.
* **Formato Profesional:** Se optimiza la presentación integrando títulos jerárquicos, etiquetas descriptivas en los ejes, una leyenda clara y una rotación de 45 grados en el eje X para garantizar que los meses no se superpongan visualmente.

```python
# ==========================================
# VISUALIZACIÓN: SERIE DE TIEMPO POR MES
# ==========================================
# Extraer el mes y año (formato YYYY-MM) de la fecha de inicio
df_anual['year_month'] = df_anual['started_at'].dt.to_period('M')

# Agrupar los datos para contar los viajes mensuales por tipo de usuario
viajes_por_mes = df_anual.groupby(['year_month', 'member_casual']).size().reset_index(name='total_viajes')
viajes_por_mes['year_month'] = viajes_por_mes['year_month'].astype(str)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(14, 7))

sns.lineplot(
    data=viajes_por_mes, 
    x='year_month', 
    y='total_viajes', 
    hue='member_casual', 
    marker='o',       
    linewidth=2.5,
    palette={'casual': '#FF8C00', 'member': '#1E90FF'} 
)

plt.title('Estacionalidad: Uso Mensual de Bicicletas (Miembros vs. Ocasionales)', fontsize=16, fontweight='bold')
plt.xlabel('Mes del Año', fontsize=12)
plt.ylabel('Cantidad Total de Viajes', fontsize=12)
plt.xticks(rotation=45) 
plt.legend(title='Tipo de Usuario', fontsize=11)
plt.tight_layout()

plt.show()
```

Una vez esclarecidos los hallazgos estas fueron las 3 recomendaciones que se hicieron en base a los resultados obtenidos:

La actividad principal de los usuarios en general se encuentra centrada los meses de junio y julio, por lo que si se quiere llegar a los usuarios lo mejor será hacerlo en esos dos meses del año.

1. Lanzar campañas de conversión enfocadas en los fines de semana:
El análisis revela que el volumen de usuarios ocasionales se dispara significativamente los sábados y domingos (alcanzando más de 440,000 viajes los sábados), en contraste con los miembros, que tienen un uso más lineal entre semana. Se debe diseñar una campaña que ataque directamente a estos usuarios de fin de semana, quizás ofreciendo una membresía anual con beneficios especiales para el ocio.

2. Crear incentivos basados en la duración del viaje y el uso de E-Bikes:
Los usuarios ocasionales realizan viajes casi el doble de largos en promedio (21.7 minutos frente a 12.4 minutos) y muestran una altísima preferencia por las bicicletas eléctricas (70%). La estrategia de marketing debe resaltar cuánto dinero ahorrarían en estos viajes largos si tuvieran una membresía anual, y se podrían ofrecer beneficios como "sin tarifa de desbloqueo en bicicletas eléctricas" al suscribirse.

3. Aprovechar la fuerte estacionalidad con promociones de primavera/verano:
Existe una caída drástica del uso entre diciembre y febrero, seguida de un repunte masivo a partir de mayo que culmina en julio/agosto. Las campañas para convertir usuarios ocasionales a miembros deben ser agresivas y lanzarse justo al inicio de la primavera, captando a los usuarios en el momento en que su intención de uso natural está en su punto máximo.


## ¿Cómo podria la empresa aprovechar estas recomendaciones?

* Campañas hiperdirigidas: Diseñar anuncios en redes sociales que se activen específicamente los viernes por la tarde y fines de semana, promocionando la conveniencia de la membresía anual para planes de ocio.

* Mensajes basados en el ahorro: Enviar correos electrónicos o notificaciones personalizadas a los usuarios ocasionales recurrentes, comparando lo que gastaron en sus viajes largos (más de 20 minutos) en bicicletas eléctricas frente a lo que habrían pagado siendo miembros.

* Sincronización estacional: Reservar la mayor parte del presupuesto publicitario de conversión para lanzarlo entre abril y mayo, interceptando a los usuarios justo cuando su intención natural de alquiler comienza a subir hacia el pico de verano.
